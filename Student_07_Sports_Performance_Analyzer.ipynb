## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# Select sports performance columns
sports_data = df[["Running_Points",
                  "Jump_Points",
                  "Throw_Points"]]

# Convert Pandas data into NumPy array
points_array = sports_data.to_numpy()

# Calculate average points
average_points = np.mean(points_array)

# Calculate maximum points
maximum_points = np.max(points_array)

# Calculate minimum points
minimum_points = np.min(points_array)

# Display results
print("Average Points:", average_points)
print("Maximum Points:", maximum_points)
print("Minimum Points:", minimum_points)
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# Calculate total points of each participant
df["Total_Points"] = (
    df["Running_Points"] +
    df["Jump_Points"] +
    df["Throw_Points"]
)

# Calculate average points of each participant
df["Average_Points"] = df["Total_Points"] / 3

# Filter participants scoring more than 70 total points
high_performers = df[df["Total_Points"] > 70]

print("Participants scoring more than 70 points:")
display(high_performers)

# Sort participants by total points
sorted_data = df.sort_values(
    by="Total_Points",
    ascending=False
)

print("Top 5 performers:")
display(sorted_data.head(5))

print("Bottom 5 performers:")
display(sorted_data.tail(5))
```

## 6. Your individual project challenge
**Challenge:** Calculate total points; rank participants; identify strongest event by average points.

Write your solution below and explain what the result means.


```python
# Calculate total points
df["Total_Points"] = (
    df["Running_Points"] +
    df["Jump_Points"] +
    df["Throw_Points"]
)

# Rank participants based on total points
df["Rank"] = df["Total_Points"].rank(
    ascending=False,
    method="min"
).astype(int)

# Sort participants according to rank
df = df.sort_values(by="Rank")

print("Sports Performance Ranking:")
display(df[["Student_ID", "Name", "Total_Points", "Rank"]])

# Calculate average points in each event
running_average = np.mean(df["Running_Points"])
jump_average = np.mean(df["Jump_Points"])
throw_average = np.mean(df["Throw_Points"])

print("\nAverage points in each event:")
print("Running:", running_average)
print("Jump:", jump_average)
print("Throw:", throw_average)

# Identify the strongest event
event_averages = {
    "Running": running_average,
    "Jump": jump_average,
    "Throw": throw_average
}

strongest_event = max(event_averages, key=event_averages.get)

print("\nStrongest Sports Event:", strongest_event)
print("Average Points:", event_averages[strongest_event])

# Identify the highest-performing participant
top_student = df.iloc[0]

print("\nTop Performing Participant:")
print("Name:", top_student["Name"])
print("Total Points:", top_student["Total_Points"])
print("Rank:", top_student["Rank"])
```
